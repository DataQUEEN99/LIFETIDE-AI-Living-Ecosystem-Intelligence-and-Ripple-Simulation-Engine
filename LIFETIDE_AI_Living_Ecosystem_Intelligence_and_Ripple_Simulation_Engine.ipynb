```python
# ================================================================
# LIFETIDE AI
# Living Ecosystem Intelligence and Ripple Simulation Engine
#
# COMPLETE ONE-CELL NOTEBOOK VERSION
# NO STREAMLIT
# NO SEPARATE APPLICATION
# NO EXTERNAL WEB SERVER
# ================================================================


# ================================================================
# 1. INSTALL AND IMPORT DEPENDENCIES
# ================================================================

import sys
import subprocess
import importlib.util
import warnings

warnings.filterwarnings("ignore")


required_packages = {
    "numpy": "numpy",
    "pandas": "pandas",
    "plotly": "plotly",
    "networkx": "networkx",
    "sklearn": "scikit-learn",
    "ipywidgets": "ipywidgets"
}


for module_name, package_name in required_packages.items():

    if importlib.util.find_spec(module_name) is None:

        subprocess.check_call(
            [
                sys.executable,
                "-m",
                "pip",
                "install",
                "-q",
                package_name
            ]
        )


import numpy as np
import pandas as pd
import networkx as nx

import plotly.graph_objects as go
import plotly.express as px

from sklearn.ensemble import RandomForestRegressor

from sklearn.model_selection import train_test_split

from sklearn.metrics import (
    mean_absolute_error,
    r2_score
)

from IPython.display import (
    display,
    HTML,
    clear_output
)

import ipywidgets as widgets


print("LIFETIDE AI dependencies are ready.")


# ================================================================
# 2. PROJECT HEADER
# ================================================================

display(
    HTML(
        """
        <div style="
            background: linear-gradient(
                135deg,
                #061b2e,
                #075766
            );
            padding: 35px;
            border-radius: 24px;
            color: white;
            font-family: Arial;
            margin-bottom: 25px;
        ">

            <h1 style="
                font-size: 48px;
                margin: 0;
            ">
                LIFETIDE AI
            </h1>

            <h2 style="
                font-weight: 400;
            ">
                Living Ecosystem Intelligence and Ripple Simulation Engine
            </h2>

            <p style="
                font-size: 20px;
            ">
                Change one thing. Watch life respond.
            </p>

            <p>
                Graph Intelligence | Ecosystem Simulation |
                Machine Learning | Intervention Analysis
            </p>

        </div>
        """
    )
)


# ================================================================
# 3. ECOSYSTEM KNOWLEDGE BASE
# ================================================================

SPECIES = {

    "Coral": {
        "type": "Foundation",
        "base": 86,
        "temperature": 1.8,
        "pollution": 1.4,
        "ph": 1.4
    },

    "Algae": {
        "type": "Producer",
        "base": 80,
        "temperature": 0.8,
        "pollution": 0.8,
        "ph": 0.6
    },

    "Zooplankton": {
        "type": "Consumer",
        "base": 77,
        "temperature": 0.6,
        "pollution": 0.8,
        "ph": 0.5
    },

    "Small Fish": {
        "type": "Consumer",
        "base": 75,
        "temperature": 0.7,
        "pollution": 1.0,
        "ph": 0.4
    },

    "Herbivorous Fish": {
        "type": "Consumer",
        "base": 78,
        "temperature": 0.8,
        "pollution": 0.9,
        "ph": 0.5
    },

    "Predatory Fish": {
        "type": "Predator",
        "base": 70,
        "temperature": 0.9,
        "pollution": 1.1,
        "ph": 0.3
    },

    "Sea Turtle": {
        "type": "Consumer",
        "base": 74,
        "temperature": 0.7,
        "pollution": 1.5,
        "ph": 0.3
    },

    "Crab": {
        "type": "Consumer",
        "base": 71,
        "temperature": 0.8,
        "pollution": 1.2,
        "ph": 0.4
    },

    "Bacteria": {
        "type": "Microbial",
        "base": 79,
        "temperature": 0.5,
        "pollution": 0.6,
        "ph": 1.0
    },

    "Seagrass": {
        "type": "Producer",
        "base": 81,
        "temperature": 0.6,
        "pollution": 0.8,
        "ph": 0.5
    },

    "Jellyfish": {
        "type": "Consumer",
        "base": 65,
        "temperature": 0.4,
        "pollution": 0.5,
        "ph": 0.2
    },

    "Shark": {
        "type": "Top Predator",
        "base": 62,
        "temperature": 0.7,
        "pollution": 1.3,
        "ph": 0.2
    }
}


# ================================================================
# 4. ECOSYSTEM RELATIONSHIP GRAPH
# ================================================================

EDGES = [

    ("Coral", "Algae"),

    ("Algae", "Herbivorous Fish"),

    ("Algae", "Zooplankton"),

    ("Bacteria", "Zooplankton"),

    ("Zooplankton", "Small Fish"),

    ("Small Fish", "Predatory Fish"),

    ("Herbivorous Fish", "Predatory Fish"),

    ("Predatory Fish", "Shark"),

    ("Small Fish", "Shark"),

    ("Seagrass", "Sea Turtle"),

    ("Seagrass", "Herbivorous Fish"),

    ("Crab", "Predatory Fish"),

    ("Jellyfish", "Predatory Fish"),

    ("Jellyfish", "Shark")
]


ecosystem_graph = nx.DiGraph()


for species_name in SPECIES:

    ecosystem_graph.add_node(
        species_name
    )


for source, target in EDGES:

    ecosystem_graph.add_edge(
        source,
        target
    )


# ================================================================
# 5. SYNTHETIC TRAINING DATA
# ================================================================

np.random.seed(42)

N = 2500


training_data = pd.DataFrame({

    "temperature":
        np.random.uniform(
            -3,
            5,
            N
        ),

    "pollution":
        np.random.uniform(
            0,
            100,
            N
        ),

    "ph_stress":
        np.random.uniform(
            0,
            5,
            N
        ),

    "resources":
        np.random.uniform(
            -50,
            50,
            N
        ),

    "disturbance":
        np.random.uniform(
            0,
            100,
            N
        )
})


training_target = (

    abs(
        training_data["temperature"]
    ) * 9

    +

    training_data["pollution"] * 0.42

    +

    training_data["ph_stress"] * 8

    -

    training_data["resources"] * 0.14

    +

    training_data["disturbance"] * 0.30

    +

    np.random.normal(
        0,
        4,
        N
    )
)


training_target = np.clip(
    training_target,
    0,
    100
)


# ================================================================
# 6. MACHINE LEARNING MODEL
# ================================================================

X_train, X_test, y_train, y_test = train_test_split(

    training_data,

    training_target,

    test_size=0.20,

    random_state=42
)


risk_model = RandomForestRegressor(

    n_estimators=180,

    max_depth=12,

    random_state=42,

    n_jobs=-1
)


risk_model.fit(
    X_train,
    y_train
)


test_prediction = risk_model.predict(
    X_test
)


model_mae = mean_absolute_error(
    y_test,
    test_prediction
)


model_r2 = r2_score(
    y_test,
    test_prediction
)


# ================================================================
# 7. HEALTH SIMULATION FUNCTION
# ================================================================

def calculate_health(
    data,
    temperature,
    pollution,
    ph_stress,
    resources,
    disturbance
):

    temperature_penalty = (

        abs(temperature)

        *

        data["temperature"]

        *

        4
    )


    pollution_penalty = (

        pollution / 10

        *

        data["pollution"]

        *

        2
    )


    ph_penalty = (

        ph_stress

        *

        data["ph"]

        *

        3
    )


    resource_effect = (

        resources

        *

        0.15
    )


    disturbance_penalty = (

        disturbance

        *

        0.08
    )


    health = (

        data["base"]

        -

        temperature_penalty

        -

        pollution_penalty

        -

        ph_penalty

        +

        resource_effect

        -

        disturbance_penalty
    )


    return float(
        np.clip(
            health,
            0,
            100
        )
    )


# ================================================================
# 8. RIPPLE ENGINE
# ================================================================

def calculate_ripple(
    dataframe,
    graph
):

    health_map = dict(
        zip(
            dataframe["Species"],
            dataframe["Health"]
        )
    )


    ripple = {}


    for node in graph.nodes:

        upstream_nodes = list(
            graph.predecessors(
                node
            )
        )


        if len(upstream_nodes) == 0:

            ripple[node] = (
                health_map[node]
            )


        else:

            upstream_health = np.mean(

                [
                    health_map[item]
                    for item in upstream_nodes
                ]
            )


            ripple[node] = (

                health_map[node] * 0.65

                +

                upstream_health * 0.35
            )


    return ripple


# ================================================================
# 9. ECOSYSTEM METRICS
# ================================================================

def calculate_metrics(
    dataframe,
    graph
):

    mean_health = float(
        dataframe["Health"].mean()
    )


    stress = (
        100 -
        mean_health
    )


    graph_density = (

        nx.density(graph)

        *

        100
    )


    stability = (

        mean_health * 0.75

        +

        mean_health * 0.15

        +

        graph_density * 0.10
    )


    return {

        "biodiversity":
            float(
                np.clip(
                    mean_health,
                    0,
                    100
                )
            ),

        "stress":
            float(
                np.clip(
                    stress,
                    0,
                    100
                )
            ),

        "stability":
            float(
                np.clip(
                    stability,
                    0,
                    100
                )
            )
    }


# ================================================================
# 10. BASELINE ECOSYSTEM
# ================================================================

baseline_rows = []


for species_name, data in SPECIES.items():

    baseline_rows.append({

        "Species":
            species_name,

        "Type":
            data["type"],

        "Health":
            data["base"]
    })


baseline_df = pd.DataFrame(
    baseline_rows
)


baseline_metrics = calculate_metrics(
    baseline_df,
    ecosystem_graph
)


# ================================================================
# 11. INTERACTIVE CONTROLS
# ================================================================

temperature_widget = widgets.FloatSlider(

    value=0.0,

    min=-3.0,

    max=5.0,

    step=0.1,

    description="Temperature:",

    continuous_update=False,

    style={
        "description_width":
        "initial"
    },

    layout=widgets.Layout(
        width="500px"
    )
)


pollution_widget = widgets.FloatSlider(

    value=10.0,

    min=0,

    max=100,

    step=1,

    description="Pollution:",

    continuous_update=False,

    style={
        "description_width":
        "initial"
    },

    layout=widgets.Layout(
        width="500px"
    )
)


ph_widget = widgets.FloatSlider(

    value=0.0,

    min=0,

    max=5,

    step=0.1,

    description="pH Stress:",

    continuous_update=False,

    style={
        "description_width":
        "initial"
    },

    layout=widgets.Layout(
        width="500px"
    )
)


resources_widget = widgets.FloatSlider(

    value=0.0,

    min=-50,

    max=50,

    step=1,

    description="Resource Availability:",

    continuous_update=False,

    style={
        "description_width":
        "initial"
    },

    layout=widgets.Layout(
        width="500px"
    )
)


disturbance_widget = widgets.FloatSlider(

    value=10.0,

    min=0,

    max=100,

    step=1,

    description="Ecosystem Disturbance:",

    continuous_update=False,

    style={
        "description_width":
        "initial"
    },

    layout=widgets.Layout(
        width="500px"
    )
)


intervention_widget = widgets.Dropdown(

    options=[

        "Reduce Pollution",

        "Reduce Temperature Stress",

        "Restore Resources",

        "Reduce Disturbance"
    ],

    value="Reduce Pollution",

    description="Intervention:",

    style={
        "description_width":
        "initial"
    },

    layout=widgets.Layout(
        width="500px"
    )
)


# ================================================================
# 12. OUTPUT CONTAINER
# ================================================================

output = widgets.Output()


# ================================================================
# 13. MAIN LIFETIDE ENGINE
# ================================================================

def run_lifetide(

    temperature,

    pollution,

    ph_stress,

    resources,

    disturbance,

    intervention
):

    with output:

        clear_output(
            wait=True
        )


        # ========================================================
        # CURRENT ECOSYSTEM STATE
        # ========================================================

        current_rows = []


        for species_name, data in SPECIES.items():

            current_rows.append({

                "Species":
                    species_name,

                "Type":
                    data["type"],

                "Health":
                    calculate_health(

                        data,

                        temperature,

                        pollution,

                        ph_stress,

                        resources,

                        disturbance
                    )
            })


        current_df = pd.DataFrame(
            current_rows
        )


        # ========================================================
        # RIPPLE ANALYSIS
        # ========================================================

        ripple = calculate_ripple(

            current_df,

            ecosystem_graph
        )


        current_df[
            "Ripple Health"
        ] = (

            current_df[
                "Species"
            ]

            .map(
                ripple
            )
        )


        current_df[
            "Stress"
        ] = (

            100

            -

            current_df[
                "Ripple Health"
            ]
        )


        # ========================================================
        # ECOSYSTEM METRICS
        # ========================================================

        current_metrics = calculate_metrics(

            current_df,

            ecosystem_graph
        )


        stability_delta = (

            current_metrics[
                "stability"
            ]

            -

            baseline_metrics[
                "stability"
            ]
        )


        biodiversity_delta = (

            current_metrics[
                "biodiversity"
            ]

            -

            baseline_metrics[
                "biodiversity"
            ]
        )


        weakest = current_df.loc[

            current_df[
                "Ripple Health"
            ].idxmin()
        ]


        # ========================================================
        # ML RISK PREDICTION
        # ========================================================

        current_features = pd.DataFrame({

            "temperature":
                [temperature],

            "pollution":
                [pollution],

            "ph_stress":
                [ph_stress],

            "resources":
                [resources],

            "disturbance":
                [disturbance]
        })


        predicted_risk = float(

            risk_model.predict(
                current_features
            )[0]
        )


        # ========================================================
        # INTERVENTION
        # ========================================================

        intervention_temperature = (
            temperature
        )

        intervention_pollution = (
            pollution
        )

        intervention_resources = (
            resources
        )

        intervention_disturbance = (
            disturbance
        )


        if intervention == "Reduce Pollution":

            intervention_pollution = max(

                0,

                pollution - 30
            )


        elif intervention == "Reduce Temperature Stress":

            intervention_temperature = (

                temperature - 1
            )


        elif intervention == "Restore Resources":

            intervention_resources = (

                resources + 25
            )


        elif intervention == "Reduce Disturbance":

            intervention_disturbance = max(

                0,

                disturbance - 30
            )


        # ========================================================
        # INTERVENTION ECOSYSTEM
        # ========================================================

        intervention_rows = []


        for species_name, data in SPECIES.items():

            intervention_rows.append({

                "Species":
                    species_name,

                "Type":
                    data["type"],

                "Health":
                    calculate_health(

                        data,

                        intervention_temperature,

                        intervention_pollution,

                        ph_stress,

                        intervention_resources,

                        intervention_disturbance
                    )
            })


        intervention_df = pd.DataFrame(
            intervention_rows
        )


        intervention_metrics = calculate_metrics(

            intervention_df,

            ecosystem_graph
        )


        intervention_change = (

            intervention_metrics[
                "stability"
            ]

            -

            current_metrics[
                "stability"
            ]
        )


        # ========================================================
        # APPLICATION HEADER
        # ========================================================

        display(

            HTML(
                """
                <div style="
                    background:
                    linear-gradient(
                        135deg,
                        #061b2e,
                        #075766
                    );
                    padding: 28px;
                    border-radius: 20px;
                    color: white;
                    margin-bottom: 20px;
                ">

                <h2>
                    LIFETIDE AI - Live Ecosystem Simulation
                </h2>

                <p>
                    Change one environmental variable and
                    observe the simulated ripple across the
                    ecosystem network.
                </p>

                </div>
                """
            )
        )


        # ========================================================
        # METRIC CARDS
        # ========================================================

        metric_html = f"""

        <div style="
            display:grid;
            grid-template-columns:
            repeat(4, 1fr);
            gap:15px;
            margin-bottom:25px;
        ">

            <div style="
                background:#10263b;
                padding:22px;
                border-radius:18px;
                color:white;
                text-align:center;
            ">

                <h1>
                    {current_metrics["stability"]:.1f}
                </h1>

                <p>
                    ECOSYSTEM STABILITY
                </p>

            </div>


            <div style="
                background:#10263b;
                padding:22px;
                border-radius:18px;
                color:white;
                text-align:center;
            ">

                <h1>
                    {current_metrics["biodiversity"]:.1f}
                </h1>

                <p>
                    BIODIVERSITY INDEX
                </p>

            </div>


            <div style="
                background:#10263b;
                padding:22px;
                border-radius:18px;
                color:white;
                text-align:center;
            ">

                <h1>
                    {current_metrics["stress"]:.1f}
                </h1>

                <p>
                    SYSTEM STRESS
                </p>

            </div>


            <div style="
                background:#10263b;
                padding:22px;
                border-radius:18px;
                color:white;
                text-align:center;
            ">

                <h2>
                    {weakest["Species"]}
                </h2>

                <p>
                    MOST VULNERABLE NODE
                </p>

            </div>

        </div>
        """


        display(
            HTML(
                metric_html
            )
        )


        # ========================================================
        # AI INTERPRETATION
        # ========================================================

        if stability_delta < -15:

            interpretation = (

                "The simulated ecosystem shows strong "
                "degradation. The disturbance propagates "
                "through several interconnected nodes."
            )


        elif stability_delta < -5:

            interpretation = (

                "The simulated ecosystem shows moderate "
                "degradation with multiple stressed nodes."
            )


        elif stability_delta < 0:

            interpretation = (

                "The ecosystem shows measurable stress "
                "but remains relatively stable under the "
                "selected scenario."
            )


        else:

            interpretation = (

                "The ecosystem remains relatively stable "
                "under the selected scenario."
            )


        display(

            HTML(
                f"""

                <div style="
                    background:#07353f;
                    padding:22px;
                    border-radius:18px;
                    color:white;
                    border-left:
                    5px solid #42d7d0;
                    margin-bottom:20px;
                ">

                    <h3>
                        Ecosystem Analysis
                    </h3>

                    <p>
                        {interpretation}
                    </p>

                    <p>
                        Biodiversity change:
                        <strong>
                            {biodiversity_delta:.2f}
                        </strong>
                    </p>

                    <p>
                        Stability change:
                        <strong>
                            {stability_delta:.2f}
                        </strong>
                    </p>

                    <p>
                        Highest vulnerability:
                        <strong>
                            {weakest["Species"]}
                        </strong>
                    </p>

                    <p>
                        Simulated ML risk:
                        <strong>
                            {predicted_risk:.2f}%
                        </strong>
                    </p>

                </div>

                """
            )
        )


        # ========================================================
        # BASELINE VS CURRENT
        # ========================================================

        comparison = pd.DataFrame({

            "Metric": [

                "Stability",

                "Biodiversity",

                "Stress"
            ],

            "Baseline": [

                baseline_metrics[
                    "stability"
                ],

                baseline_metrics[
                    "biodiversity"
                ],

                baseline_metrics[
                    "stress"
                ]
            ],

            "Simulation": [

                current_metrics[
                    "stability"
                ],

                current_metrics[
                    "biodiversity"
                ],

                current_metrics[
                    "stress"
                ]
            ]
        })


        comparison_fig = go.Figure()


        comparison_fig.add_trace(

            go.Bar(

                name="Baseline",

                x=comparison[
                    "Metric"
                ],

                y=comparison[
                    "Baseline"
                ]
            )
        )


        comparison_fig.add_trace(

            go.Bar(

                name="Simulation",

                x=comparison[
                    "Metric"
                ],

                y=comparison[
                    "Simulation"
                ]
            )
        )


        comparison_fig.update_layout(

            title="Baseline versus Current Ecosystem",

            barmode="group",

            template="plotly_dark",

            height=450
        )


        comparison_fig.show()


        # ========================================================
        # SPECIES HEALTH
        # ========================================================

        health_plot = current_df.sort_values(

            "Ripple Health"
        )


        health_fig = px.bar(

            health_plot,

            x="Ripple Health",

            y="Species",

            color="Ripple Health",

            orientation="h",

            template="plotly_dark",

            title="Life Ripple Health"
        )


        health_fig.update_layout(

            height=600
        )


        health_fig.show()


        # ========================================================
        # NETWORK GRAPH
        # ========================================================

        positions = nx.spring_layout(

            ecosystem_graph,

            seed=42,

            k=1.5
        )


        edge_x = []

        edge_y = []


        for source, target in ecosystem_graph.edges:

            x0, y0 = positions[source]

            x1, y1 = positions[target]


            edge_x.extend(

                [
                    x0,
                    x1,
                    None
                ]
            )


            edge_y.extend(

                [
                    y0,
                    y1,
                    None
                ]
            )


        edge_trace = go.Scatter(

            x=edge_x,

            y=edge_y,

            mode="lines",

            line=dict(
                width=1
            ),

            hoverinfo="none"
        )


        node_x = []

        node_y = []

        node_color = []

        node_text = []


        for node in ecosystem_graph.nodes:

            x, y = positions[node]


            node_x.append(x)

            node_y.append(y)


            health_value = ripple[node]


            node_color.append(
                health_value
            )


            node_text.append(

                f"{node}<br>"
                f"Ripple Health: "
                f"{health_value:.2f}"
            )


        node_trace = go.Scatter(

            x=node_x,

            y=node_y,

            mode="markers+text",

            text=list(
                ecosystem_graph.nodes
            ),

            textposition="top center",

            hovertext=node_text,

            hoverinfo="text",

            marker=dict(

                size=27,

                color=node_color,

                colorscale="Turbo",

                showscale=True,

                colorbar=dict(
                    title="Health"
                )
            )
        )


        network_fig = go.Figure(

            data=[
                edge_trace,
                node_trace
            ]
        )


        network_fig.update_layout(

            title=
            "Living Ecosystem Relationship Network",

            template="plotly_dark",

            height=700,

            showlegend=False,

            xaxis=dict(
                visible=False
            ),

            yaxis=dict(
                visible=False
            )
        )


        network_fig.show()


        # ========================================================
        # VULNERABILITY ANALYSIS
        # ========================================================

        vulnerability = current_df[

            [
                "Species",

                "Type",

                "Ripple Health",

                "Stress"
            ]

        ].copy()


        def classify_risk(
            value
        ):

            if value >= 60:

                return "CRITICAL"

            elif value >= 40:

                return "HIGH"

            elif value >= 20:

                return "MODERATE"

            else:

                return "LOW"


        vulnerability[
            "Risk Level"
        ] = (

            vulnerability[
                "Stress"
            ]

            .apply(
                classify_risk
            )
        )


        vulnerability = vulnerability.sort_values(

            "Stress",

            ascending=False
        )


        display(

            HTML(
                "<h2>Vulnerability Intelligence</h2>"
            )
        )


        display(

            vulnerability.style
            .background_gradient(

                subset=[
                    "Stress"
                ],

                cmap="RdYlGn_r"
            )
        )


        # ========================================================
        # INTERVENTION ANALYSIS
        # ========================================================

        display(

            HTML(
                "<h2>What-If Intervention Lab</h2>"
            )
        )


        intervention_comparison = pd.DataFrame({

            "State": [

                "Current",

                "Intervention"
            ],

            "Stability": [

                current_metrics[
                    "stability"
                ],

                intervention_metrics[
                    "stability"
                ]
            ],

            "Biodiversity": [

                current_metrics[
                    "biodiversity"
                ],

                intervention_metrics[
                    "biodiversity"
                ]
            ]
        })


        intervention_fig = go.Figure()


        intervention_fig.add_trace(

            go.Bar(

                name="Stability",

                x=intervention_comparison[
                    "State"
                ],

                y=intervention_comparison[
                    "Stability"
                ]
            )
        )


        intervention_fig.add_trace(

            go.Bar(

                name="Biodiversity",

                x=intervention_comparison[
                    "State"
                ],

                y=intervention_comparison[
                    "Biodiversity"
                ]
            )
        )


        intervention_fig.update_layout(

            title=
            f"Intervention Analysis: {intervention}",

            barmode="group",

            template="plotly_dark",

            height=450
        )


        intervention_fig.show()


        display(

            HTML(
                f"""

                <div style="
                    background:#07353f;
                    padding:20px;
                    border-radius:18px;
                    color:white;
                    margin-top:15px;
                ">

                    <h3>
                        Intervention Result
                    </h3>

                    <p>
                        Selected intervention:
                        <strong>
                            {intervention}
                        </strong>
                    </p>

                    <p>
                        Simulated stability change:
                        <strong>
                            {intervention_change:+.2f}
                        </strong>
                        points
                    </p>

                </div>

                """
            )
        )


        # ========================================================
        # FEATURE IMPORTANCE
        # ========================================================

        importance = pd.DataFrame({

            "Factor":
                training_data.columns,

            "Importance":
                risk_model.feature_importances_
        })


        importance = importance.sort_values(

            "Importance",

            ascending=True
        )


        importance_fig = px.bar(

            importance,

            x="Importance",

            y="Factor",

            orientation="h",

            template="plotly_dark",

            title="Machine Learning Feature Importance"
        )


        importance_fig.update_layout(

            height=450
        )


        importance_fig.show()


        # ========================================================
        # FUTURE PROJECTION
        # ========================================================

        years = np.arange(

            0,

            11
        )


        future_stability = []


        for year in years:

            deterioration = (

                predicted_risk

                *

                0.035

                *

                year
            )


            recovery = (

                max(
                    intervention_change,
                    0
                )

                *

                0.025

                *

                year
            )


            projected_value = (

                current_metrics[
                    "stability"
                ]

                -

                deterioration

                +

                recovery
            )


            future_stability.append(

                np.clip(

                    projected_value,

                    0,

                    100
                )
            )


        future_df = pd.DataFrame({

            "Year":
                years,

            "Projected Stability":
                future_stability
        })


        future_fig = px.line(

            future_df,

            x="Year",

            y="Projected Stability",

            markers=True,

            template="plotly_dark",

            title=
            "Simulated Ten-Year Ecosystem Projection"
        )


        future_fig.update_layout(

            height=450
        )


        future_fig.show()


        # ========================================================
        # MACHINE LEARNING PERFORMANCE
        # ========================================================

        display(

            HTML(
                f"""

                <h2>
                    Machine Learning Model
                </h2>

                <div style="
                    display:flex;
                    gap:20px;
                    margin-bottom:20px;
                ">

                    <div style="
                        background:#10263b;
                        color:white;
                        padding:20px;
                        border-radius:15px;
                        flex:1;
                    ">

                        <h2>
                            {predicted_risk:.2f}%
                        </h2>

                        <p>
                            Current Simulated Risk
                        </p>

                    </div>


                    <div style="
                        background:#10263b;
                        color:white;
                        padding:20px;
                        border-radius:15px;
                        flex:1;
                    ">

                        <h2>
                            {model_mae:.2f}
                        </h2>

                        <p>
                            Mean Absolute Error
                        </p>

                    </div>


                    <div style="
                        background:#10263b;
                        color:white;
                        padding:20px;
                        border-radius:15px;
                        flex:1;
                    ">

                        <h2>
                            {model_r2:.2f}
                        </h2>

                        <p>
                            R Squared
                        </p>

                    </div>

                </div>

                """
            )
        )


        # ========================================================
        # DATA EXPORT
        # ========================================================

        export_df = current_df.copy()


        export_df[
            "Ecosystem Stability"
        ] = (

            current_metrics[
                "stability"
            ]
        )


        export_df[
            "Biodiversity Index"
        ] = (

            current_metrics[
                "biodiversity"
            ]
        )


        export_df[
            "System Stress"
        ] = (

            current_metrics[
                "stress"
            ]
        )


        export_df[
            "Simulated Risk"
        ] = predicted_risk


        export_df[
            "Intervention"
        ] = intervention


        csv_data = export_df.to_csv(

            index=False
        )


        with open(

            "LIFETIDE_AI_Ecosystem_Analysis.csv",

            "w",

            encoding="utf-8"

        ) as file:

            file.write(
                csv_data
            )


        display(

            HTML(
                """
                <h2>
                    Analysis Export
                </h2>

                <div style="
                    background:#10263b;
                    color:white;
                    padding:18px;
                    border-radius:15px;
                ">

                    Analysis saved as:

                    <strong>
                        LIFETIDE_AI_Ecosystem_Analysis.csv
                    </strong>

                </div>
                """
            )
        )


        # ========================================================
        # FINAL PROJECT SUMMARY
        # ========================================================

        display(

            HTML(
                """
                <div style="
                    background:
                    linear-gradient(
                        135deg,
                        #061b2e,
                        #075766
                    );

                    padding:30px;

                    border-radius:20px;

                    color:white;

                    margin-top:25px;
                ">

                    <h2>
                        LIFETIDE AI
                    </h2>

                    <p>
                        LIFETIDE AI combines graph intelligence,
                        ecosystem simulation, machine learning,
                        vulnerability analysis and intervention
                        modelling into a computational research
                        prototype.
                    </p>

                    <p>
                        Graph -> Ripple -> Risk ->
                        Intervention -> Future
                    </p>

                </div>
                """
            )
        )


        # ========================================================
        # DISCLAIMER
        # ========================================================

        display(

            HTML(
                """
                <div style="
                    background:#241b0c;
                    color:#ffd88a;
                    padding:20px;
                    border-radius:15px;
                    margin-top:20px;
                ">

                    <strong>
                        Research and Hackathon Disclaimer
                    </strong>

                    <p>
                        LIFETIDE AI is a computational prototype.
                        The ecosystem values, relationships,
                        risk scores and future projections are
                        simulated or model-defined outputs.
                        They are not validated ecological
                        measurements, conservation recommendations,
                        or scientific forecasts.
                    </p>

                </div>
                """
            )
        )


# ================================================================
# 14. CONTROL PANEL
# ================================================================

controls = widgets.VBox(

    [

        widgets.HTML(
            "<h2>Life Ripple Control Panel</h2>"
        ),

        temperature_widget,

        pollution_widget,

        ph_widget,

        resources_widget,

        disturbance_widget,

        widgets.HTML(
            "<hr><h3>Intervention Lab</h3>"
        ),

        intervention_widget
    ],

    layout=widgets.Layout(
        padding="15px"
    )
)


# ================================================================
# 15. CONNECT CONTROLS TO ENGINE
# ================================================================

interactive_output = widgets.interactive_output(

    run_lifetide,

    {

        "temperature":
            temperature_widget,

        "pollution":
            pollution_widget,

        "ph_stress":
            ph_widget,

        "resources":
            resources_widget,

        "disturbance":
            disturbance_widget,

        "intervention":
            intervention_widget
    }
)


# ================================================================
# 16. DISPLAY APPLICATION
# ================================================================

display(
    controls
)

display(
    output
)


# ================================================================
# 17. INITIAL SIMULATION
# ================================================================

run_lifetide(

    temperature_widget.value,

    pollution_widget.value,

    ph_widget.value,

    resources_widget.value,

    disturbance_widget.value,

    intervention_widget.value
)


# ================================================================
# 18. COMPLETION MESSAGE
# ================================================================

print()
print("=" * 70)
print("LIFETIDE AI COMPLETE")
print("=" * 70)
print("Streamlit: Not used")
print("Application type: Notebook")
print("Execution mode: Single cell")
print("Interactive controls: Enabled")
print("Graph simulation: Enabled")
print("Machine learning: Enabled")
print("Intervention analysis: Enabled")
print("Future projection: Enabled")
print("CSV export: Enabled")
print("=" * 70)
```
